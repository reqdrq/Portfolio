## #03. 데이터 품질 점검

### 1. 결측치 확인

```python
# 보유 게임수 0이면 결측치 처리
df1['games_owned'] = df1['games_owned'].replace(0, np.nan)

my_qtcheck.check_missing_values(df1)
```

### 2. 결측치 제거 

```python
df1 = df1.dropna().reset_index(drop=True)

my_qtcheck.check_missing_values(df1)
```

### 3. 윈저라이징


```python
# 1. 대상 열 지정
wtarget = ['games_owned', 'num_reviews', 'playtime_forever_min', 'playtime_at_review_min']

# 2. 각 열별 상위 기준값 계산 (Series 형태로 반환됨)
upper_bounds = df1[wtarget].quantile(0.995)

# 3. axis=1을 지정하여 각 열에 맞는 기준값으로 clip 적용
df1[wtarget] = df1[wtarget].clip(upper=upper_bounds, axis=1)

# 결과 확인
print("열별 상위 기준값:")
print(upper_bounds)
```

```python
# 파생변수 리뷰이후 플레이시간 추가
df1['playtime_after_review_min'] = df1['playtime_forever_min'] - df1['playtime_at_review_min']

# 파생변수 플레이시간 대비 업적 수
df1['mastery'] = df1['n_ach_total'] / (df1['playtime_forever_min']/60)
```

## #07. 탐색적 데이터 분석

### 1. 명목형 종속변수 단변량 분석

```python
# 범주형 분포는 카운트 플롯으로 처리
my_plot.countplot(data=df1, x=target, hue=target, palette="Set2",
                    title=f"종속변수 {target} 분포")

# 추가적으로 빈도표 생성
freq = df1[target].value_counts().rename("빈도").to_frame()
freq["비율"] = df1[target].value_counts(normalize=True)
display(freq)
```

### 2. 연속형 독립변수 단변량 분석

```python
for f in continuous_cols:
    # 히스토그램 및 박스플롯을 1행 2열 서브플롯으로 구성하는 경우
    fig, ax = my_plot.init(rows=1, cols=2, title=f"독립변수 {f} 분포",
                       width=960, height=640)
    my_plot.histplot(data=df1, x=f, kde=True, ax=ax[0])
    my_plot.boxplot(data=df1, x=f, ax=ax[1])
    my_plot.show()

    display(desc_df[desc_df.index == f])
```

### 3. T-Test (x: 연속형 독립변수 → y: 2집단 명목형 종속변수)

```python
for f in continuous_cols:
        # 빈 줄 출력용
        print()

        # 분석 단위 제목 출력
        display(Markdown(f"### ▶ {f} → {target}"))

        # 수집된 데이터가 long 타입이므로 검정 대상만 wide 타입으로 변환
        wide_df = my_prep.long2wide(df1, hue=target, values=f)
        display(my_stats.test_independent(wide_df, 
                                          group1=wide_df.columns[0], 
                                          group2=wide_df.columns[1], plot=True))
```

### 4. 교차분석 (x: 명목형 독립변수 → y: 명목형 종속변수)

```python
for n in nominal_cols:
    print()

    # 제목
    display(Markdown(f"### ▶ {n} → {target}"))

    # 교차분석 (카이제곱 독립성 검정)
    display(my_stats.chi2_independence(df1, x=n, y=target, plot=True))
```

### 5. 독립변수간 상관분석

```python
corr_cols = ['games_owned', 'num_reviews', 'playtime_forever_min', 'playtime_at_review_min', 'stage_final', 'n_ach_total', 'stage_at_review', 'n_ach_at_review']

corr = my_stats.multi_correlation(df1, columns=corr_cols, plot=False)
display(my_stats.correlation_summary(corr))

# 분산팽창지수(VIF)
display(my_stats.compute_vif(df1[corr_cols]))
```
