### 3. 타입 점검 및 변환

```python
origin['gender'] = origin['gender'].astype('category')
origin['city_type'] = origin['city_type'].astype('category')
origin['doomscrolling_before_sleep'] = origin['doomscrolling_before_sleep'].astype('category')
origin['mental_state'] = origin['mental_state'].astype('category')
origin['learning_style'] = origin['learning_style'].astype('category')
origin['career_goal'] = origin['career_goal'].astype('category')
origin['performance_category'] = origin['performance_category'].astype('category')
df = origin.copy()
df = df.drop('performance_category', axis=1)
df = df.dropna().reset_index(drop=True)
df.info()
```

### 4. 변수 유형 분류

```python
# 종속변수
target = "final_exam_score"

# 명목형 변수 이름
nominal_cols = my_qtcheck.get_categorical_column_names(df)

# 연속형(종속변수 포함)
continuous_cols = my_qtcheck.get_number_column_names(df)

# 연속형 독립변수
feature_continuous = [c for c in continuous_cols if c != target]

print("목표변수 :", target)
print("명목형 :", nominal_cols)
print("연속형 :", continuous_cols)
```

## #02. 데이터 전처리

### 1. 명목형 라벨링

```python
df1 = my_prep.labeling(df, columns=nominal_cols)
df1.head()
```

### 2. 더미변수 인코딩

```python
df2 = my_prep.dummies(df1, columns=nominal_cols, drop_first=True)
df2.to_excel("student_체크포인트0_기준선.xlsx", index=False)
df2.head()
```

### 3. 로그 변환 대상 컬럼 선정

```python
# 우측꼬리 로그 변환 대상 컬럼
log_cols = []
log_cols
```

```python
# 좌측꼬리 로그 변환 대상 컬럼
reflect_cols = []
reflect_cols
```

### 4. 로그 변환


```python
df3 = my_prep.log_transform(df2, log_columns=log_cols, reflect_columns=reflect_cols)
df3.to_excel("student_체크포인트1_로그변환.xlsx", index=False)
df3.head()
```

### 5. 이상치 대체


```python
df4 = my_prep.replace_outlier(df3, columns=feature_continuous)
df4.to_excel("student_체크포인트2_이상치대체.xlsx", index=False)
df4.head()
```

### 6. 다중 공선성 제거

```python
df5 = my_prep.reduce_vif(df4, columns=feature_continuous, threshold=10.0)
df5.to_excel("student_체크포인트3_다중공선성제거.xlsx", index=False)
df5.head()
```

### 7. 정규화

```python
df6 = my_prep.scaling(df5, columns=feature_continuous, method='standard')
df6.to_excel("student_체크포인트4_정규화.xlsx", index=False)
df6.head()
```
